```python

class Student:
    def __init__(self, student_id, name, math, english, science):
        # Public attributes
        self.student_id = student_id
        self.name = name

        # Private attributes (encapsulation)
        self.__math = math
        self.__english = english
        self.__science = science

    # Calculate the student's average mark
    def calculate_average(self):
        total = self.__math + self.__english + self.__science
        return total / 3

    # Determine the student's grade
    def calculate_grade(self):
        average = self.calculate_average()

        if average >= 80:
            return "A"
        elif average >= 70:
            return "B"
        elif average >= 60:
            return "C"
        elif average >= 50:
            return "D"
        else:
            return "F"

    # Access private marks safely
    def get_marks(self):
        return {
            "Math": self.__math,
            "English": self.__english,
            "Science": self.__science
        }


# Create a Student object for Grace
grace = Student(101, "Grace", 78, 85, 80)

# Display Grace's information
print("Student ID:", grace.student_id)
print("Student Name:", grace.name)
print("Marks:", grace.get_marks())
print("Average:", round(grace.calculate_average(), 2))
print("Grade:", grace.calculate_grade())
```

```python

from abc import ABC, abstractmethod


# Abstract parent class
class StudentBase(ABC):
    def __init__(self, student_id, name, math, english, science):
        self.student_id = student_id
        self.name = name
        self.__math = math
        self.__english = english
        self.__science = science

    def calculate_average(self):
        total = self.__math + self.__english + self.__science
        return total / 3

    def calculate_grade(self):
        average = self.calculate_average()

        if average >= 80:
            return "A"
        elif average >= 70:
            return "B"
        elif average >= 60:
            return "C"
        elif average >= 50:
            return "D"
        else:
            return "F"

    @abstractmethod
    def get_student_type(self):
        pass

    @abstractmethod
    def get_benefit(self):
        pass


# Child class for regular students
class RegularStudent(StudentBase):
    def get_student_type(self):
        return "Regular Student"

    def get_benefit(self):
        return "Standard academic support"


# Child class for scholarship students
class ScholarshipStudent(StudentBase):
    def get_student_type(self):
        return "Scholarship Student"

    def get_benefit(self):
        if self.calculate_average() >= 80:
            return "Eligible to retain scholarship"
        else:
            return "Scholarship review required"


# Create objects of different student types
student1 = RegularStudent(101, "Grace", 78, 85, 80)
student2 = ScholarshipStudent(102, "Brian", 92, 88, 95)
student3 = ScholarshipStudent(104, "Kevin", 45, 55, 50)


# Demonstrate polymorphism and method overriding
students = [student1, student2, student3]

for student in students:
    print("--------------------------------")
    print("Name:", student.name)
    print("Type:", student.get_student_type())
    print("Average:", round(student.calculate_average(), 2))
    print("Grade:", student.calculate_grade())
    print("Benefit:", student.get_benefit())
```

```python
#creating a performer analysizer

import pandas as pd


class PerformanceAnalyzer:
    def __init__(self, csv_file="students.csv"):
        # Load student records from the CSV file
        self.data = pd.read_csv(csv_file)

        # Calculate each student's average mark
        subjects = ["math", "english", "science"]
        self.data["average"] = self.data[subjects].mean(axis=1)

    def highest_average(self):
        # Find the student with the highest average
        student = self.data.loc[self.data["average"].idxmax()]
        return student["name"], student["average"]

    def lowest_average(self):
        # Find the student with the lowest average
        student = self.data.loc[self.data["average"].idxmin()]
        return student["name"], student["average"]

    def class_average(self):
        # Calculate the average of all student averages
        return self.data["average"].mean()

    def display_summary(self):
        # Display the class performance summary
        highest_name, highest_mark = self.highest_average()
        lowest_name, lowest_mark = self.lowest_average()

        print("===== CLASS PERFORMANCE SUMMARY =====")
        print(f"Highest Average: {highest_name} - {highest_mark:.2f}")
        print(f"Lowest Average: {lowest_name} - {lowest_mark:.2f}")
        print(f"Overall Class Average: {self.class_average():.2f}")


# Create the analyzer object
analyzer = PerformanceAnalyzer("students.csv")

# Display the results
analyzer.display_summary()
```

```python

class FlexibleCalculator:

    # One method that accepts different numbers of marks
    def calculate_average(self, *marks):
        if len(marks) == 0:
            return 0

        return sum(marks) / len(marks)


# Create a calculator object
calculator = FlexibleCalculator()

# Call the same method with different numbers of arguments
print("Average of two marks:", calculator.calculate_average(80, 90))
print("Average of three marks:", calculator.calculate_average(70, 80, 90))
print("Average of four marks:", calculator.calculate_average(60, 70, 80, 90))
print("Average with no marks:", calculator.calculate_average())
```

This demonstrates Python's approach to method overloading through variable-length arguments.

```python
# writing a report code

import pandas as pd

# Load the student dataset
df = pd.read_csv("students.csv")

# Calculate the average mark for each student
subjects = ["math", "english", "science"]
df["average"] = df[subjects].mean(axis=1)

# Assign grades based on average marks
def assign_grade(average):
    if average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"

df["grade"] = df["average"].apply(assign_grade)

# Determine scholarship status
def scholarship_status(row):
    if row["student_type"] != "Scholarship":
        return "Not Applicable"
    elif row["average"] >= 80:
        return "Eligible to Retain"
    else:
        return "Review Required"

df["scholarship_status"] = df.apply(
    scholarship_status, axis=1
)

# Display the complete student report
print("========== STUDENT PERFORMANCE REPORT ==========")
display(df.round({"average": 2}))

# Calculate class statistics
print("\n========== CLASS SUMMARY ==========")
print("Number of Students:", len(df))
print("Highest Average:", round(df["average"].max(), 2))
print("Lowest Average:", round(df["average"].min(), 2))
print("Overall Class Average:", round(df["average"].mean(), 2))

# Save the report to a CSV file
df.to_csv("student_performance_report.csv", index=False)

print("\nReport saved successfully!")
```
